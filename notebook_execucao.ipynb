## 5. Exemplo: Executar Scripts Python via subprocess

Demonstração de execução de scripts locais com captura de saída.

```python
def run_command(cmd: str, timeout: int = 30) -> Tuple[str, str, int]:
    """
    Executa um comando shell e retorna stdout, stderr e código de saída.
    
    Args:
        cmd: Comando a executar
        timeout: Timeout em segundos
        
    Returns:
        Tupla (stdout, stderr, returncode)
    """
    try:
        result = subprocess.run(
            cmd,
            shell=True,
            capture_output=True,
            text=True,
            timeout=timeout
        )
        return result.stdout, result.stderr, result.returncode
    except subprocess.TimeoutExpired:
        return "", f"Comando expirou após {timeout}s", -1
    except Exception as e:
        return "", str(e), -1

def run_python_script(script_path: str, args: list = None, timeout: int = 30) -> Tuple[str, str, int]:
    """
    Executa um script Python e retorna a saída.
    """
    cmd = f"{sys.executable} {script_path}"
    if args:
        cmd += " " + " ".join(args)
    return run_command(cmd, timeout)

# Teste
print("✓ Utilitários de execução definidos com sucesso")
```
